```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import pandas as pd
import re
```

```python
data = pd.read_csv("./tripadvisor_hotel_reviews.csv")
```

```python
data.info()
```

```python
data.head()
```

```python
data.Review[0]
```

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data.head()
```

```python
en_stopwords = stopwords.words('english')
en_stopwords.remove("not")
```

```python
data['review_no_stopwords'] = data['review_lowercase']\
                                .apply(lambda x: ' '.join([word for word in x.split() \
                                                    if word not in(en_stopwords)]))
```

```python
#data.head()
data['review_no_stopwords'][0]
```

```python
data['review_no_puncts'] = data.apply(lambda x: re.sub(r"[*]", r" star",x['review_no_stopwords']), axis=1)
```

```python
#data.head()
```

```python
data['review_no_puncts'] = data.apply(lambda x: re.sub(r"([^\w\s])", "", x['review_no_puncts']), axis=1)
```

```python
#data['review_no_puncts'][0]
```

```python
data['tokenize'] = data.apply(lambda x: word_tokenize(x['review_no_puncts']), axis=1)
```

```python
data.head()
```

```python
#data['tokenize'][0]
```

```python
ps = PorterStemmer()
```

```python
data["stemmed"] = data["tokenize"].apply(lambda x: [ps.stem(row) for row in x])
```

```python
data['stemmed']
```

```python
lemma = WordNetLemmatizer()
```

```python
data["lemma"]=data["tokenize"].apply(lambda x: [lemma.lemmatize(row) for row in x])
data["lemma"]
```

```python
clean_tokens = sum(data["lemma"],[])
```

```python
unigrams = (pd.Series(nltk.ngrams(clean_tokens, 1)).value_counts())
print(unigrams)
```

```python
bigrams = (pd.Series(nltk.ngrams(clean_tokens, 2)).value_counts())
print(bigrams)
```

```python
n_grams = (pd.Series(nltk.ngrams(clean_tokens, 4)).value_counts())
print(n_grams)
```
