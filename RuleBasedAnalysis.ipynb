## Rule Based Sentiment Analysis

#### using TextBlob

```python
from textblob import TextBlob
```

```python
sentence_1 = "i had great time at the movie it was really funny"
sentence_2 = "i had great time at the movie but the parking was terrible"
sentence_3 = "i had great time at the movie but the parking wasn't great"
sentence_4 = "i want to see a movie"
```

```python
print(sentence_1)
sentiment_1 = TextBlob(sentence_1)
print(sentiment_1.sentiment.polarity)
```

```python
print(sentence_2)
sentiment_2 = TextBlob(sentence_2)
print(sentiment_2.sentiment.polarity)
```

```python
print(sentence_3)
sentiment_3 = TextBlob(sentence_3)
print(sentiment_3.sentiment.polarity)
```

```python
print(sentence_4)
sentiment_4 = TextBlob(sentence_4)
print(sentiment_4.sentiment.polarity)
```

#### using Vader

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
vader_sentiment = SentimentIntensityAnalyzer()
```

```python
print(sentence_1)
print(vader_sentiment.polarity_scores(sentence_1))
```

```python
print(sentence_2)
print(vader_sentiment.polarity_scores(sentence_2))
```

```python
print(sentence_3)
print(vader_sentiment.polarity_scores(sentence_3))
```

```python
print(sentence_4)
print(vader_sentiment.polarity_scores(sentence_4))
```
